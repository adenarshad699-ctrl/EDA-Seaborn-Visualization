# Exploratory Data Analysis(Lecture  No 04)

```python
# import and setup
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style ='darkgrid')
```

## Seaborn Fundamentals

### Load Dataset

```python
df = sns.load_dataset('iris')
df.head(2)
```

```python
df = sns.load_dataset('tips')
df.head(2)
```

### Line plot

```python
sns.lineplot(x= 'total_bill', y = 'tip', data = df)
plt.show()
```

```python
plt.figure(figsize=(30,4))
sns.lineplot(x= 'total_bill', y = 'tip', data = df)
plt.show()
```

```python
plt.figure(figsize=(30,7))
sns.lineplot(x= 'total_bill', y = 'tip', data = df)
plt.title('Line Plot of Total Bill vs tip')
plt.savefig('lineplot.png')
plt.show()
```

### Barplot

```python
plt.figure(figsize=(10,4))

sns.barplot(x= 'day', y= 'total_bill', data = df)

plt.title('Bar Plot of Day wise Total Bill')

plt.savefig('barplot.png')

plt.show()
```

```python
plt.figure(figsize=(10,4))

sns.barplot(x= 'day', y= 'total_bill', data = df)

plt.title('Bar Plot of Day wise Total Bill By : Aden Arshad')

plt.savefig('barplot.png')

plt.show()
```

### Countplot

```python
plt.figure(figsize=(10,4))

sns.countplot(x = 'day', data = df)

plt.title('Count Plot Day wise sales')

plt.savefig('Countplot.png')

plt.show()
```

### Histogram

```python
plt.figure(figsize=(10,4))

sns.histplot(df['total_bill'], bins = 30)
plt.title('Frequecy Count of total Bill')

plt.savefig('histplot.png')

plt.show()
```

```python
plt.figure(figsize=(10,4))

sns.histplot(df['total_bill'], bins = 30, kde= True)
plt.title('Frequecy Count of total Bill')

plt.savefig('histplot.png')

plt.show()
```

### KDE Plot

```python
plt.figure(figsize=(10,4))

sns.kdeplot(df['total_bill'])
plt.title('Frequecy Distribution line of total Bill')

plt.savefig('kdeplot.png')

plt.show()
```

### Boxplot

```python
plt.figure(figsize=(5,5))

sns.boxplot(df['total_bill'])

plt.title('Box plot of total Bill')

plt.savefig('Boxplot1.png')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(df['tip'])

plt.title('Box plot of tips')

plt.savefig('Boxplot2.png')

plt.show()
```

```python
plt.figure(figsize=(25,5))

sns.boxplot(x = df['tip'])

plt.title('Box plot of tips')

plt.savefig('Boxplot2.png')

plt.show()
```

```python
plt.figure(figsize=(25,5))

sns.boxplot(x = df['total_bill'], y = df["day"])

plt.title('Box plot of total_bill w.r.t Day of weeks')

plt.savefig('Boxplot3.png')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.violinplot(x = df['total_bill'], y = df["time"])

plt.title('Volin plot of total_bill w.r.t Day of weeks')

plt.savefig('violinplot3.png')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.stripplot(x = df['total_bill'], y = df["time"])

plt.title('Strip plot of total_bill w.r.t Day of weeks')

plt.savefig('stripplot3.png')

plt.show()
```

### Correlation Heatmap

```python
corr = df.corr(numeric_only= True)
corr
```

```python
sns.heatmap(corr)
```

```python
sns.heatmap(corr, annot= True)
plt.title("Correlation heatmap")

plt.savefig('Correlation_heatmap.png')

plt.show()
```
